# Notebook 02 – Master Dataset Construction

## Objective

Merge all processed indicators into a single analytical dataset that will be used to construct the Climate Resilience Index for the 32 Mexican states.

## Inputs

Processed indicator datasets:

- Hazard
  - Drought Vulnerability
--------------
- Sensitivity
  - Multidimensional Poverty
  - Marginalization Index
  - Employment in the Primary Sector
--------------
- Adaptive Capacity
  - Educational Attainment

## Output

- climate_resilience_master.csv

```python
import pandas as pd
import numpy as np
```

```python
drought = pd.read_csv(
    "../data/processed/hazard_drought.csv"
)

poverty = pd.read_csv(
    "../data/processed/sensitivity_multidimensional_poverty.csv"
)

marginalization = pd.read_csv(
    "../data/processed/sensitivity_marginalization.csv"
)

employment = pd.read_csv(
    "../data/processed/sensitivity_primary_sector_employment.csv"
)

education = pd.read_csv(
    "../data/processed/adaptive_capacity_educational_attainment.csv"
)

state_lookup = pd.read_csv(
    "../data/state_codes.csv"
)
```

```python
datasets = {
    "Drought": drought,
    "Poverty": poverty,
    "Marginalization": marginalization,
    "Employment": employment,
    "Education": education
}

for name, df in datasets.items():
    print(f"\n{name}")
    print(df.shape)
    print(df.head(2))
```

```python
master = state_lookup.copy()

master.head()
```

```python
master = master.merge(
    drought,
    on=["state_code", "state_name"],
    how="left"
)
```

```python
master = master.merge(
    poverty,
    on=["state_code", "state_name"],
    how="left"
)
```

```python
master = master.merge(
    marginalization,
    on=["state_code", "state_name"],
    how="left"
)
```

```python
master = master.merge(
    employment,
    on=["state_code", "state_name"],
    how="left"
)
```

```python
master = master.merge(
    education,
    on=["state_code", "state_name"],
    how="left"
)
```

```python
hazard = pd.read_csv("../data/processed/hazard_indicators.csv")

master = master.merge(
    hazard,
    on=["state_code", "state_name"],
    how="left"
)
```

```python
master.head()
```

```python
master.columns
```

```python
master = master.drop(columns=["drought_susceptibility"])
```

```python
master.to_csv(
    "../data/processed/climate_resilience_master.csv",
    index=False
)
```

```python
master["state_code"] = (
    master["state_code"]
    .astype(str)
    .str.zfill(2)
)

master.to_csv(
    "../data/processed/climate_resilience_master.csv",
    index=False
)
```

```python
master["marginalization_score"] = (
    master["marginalization_index"].max()
    - master["marginalization_index"]
).round(2)

master.drop(
    columns=["marginalization_index"],
    inplace=True
)
```

```python
master.to_csv(
    "../data/processed/climate_resilience_master.csv",
    index=False
)
```
